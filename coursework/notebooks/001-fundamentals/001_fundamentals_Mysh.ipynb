## Ejercicio 1
Con numpy, genera un array de 50 números aleatorios entre 0 y 1.

Calcula la media, la mediana y la desviación estándar de ese array.

Redondea los resultados a dos decimales e imprime cada valor.

```python
import numpy as np
rnd = np.random.rand(50)
mean_rnd = np.round(np.mean(rnd),2)
median_rnd = np.round(np.median(rnd),2)
std_rnd= np.round(np.std(rnd),2)
print(rnd)
print("Media:", f"{mean_rnd:.2f}")
print("Mediana:", f"{median_rnd:.2f}")
print("Desviación estándar:", f"{std_rnd:.2f}")
```

##Ejercicio 4
Define una función que reciba un número entero positivo y devuelva:

"Par" si el número es par,

"Impar" si es impar,

Añade un mensaje de error si el número no es válido.

```python
def Chi(a):
    try:
        num = int(a)
        if num <= 0:
            print("El numero no es valido (debe ser un entero positivo)")
        elif num % 2 == 1:
            print("Impar")
        else:
            print("Par")
    except ValueError:
        print("El numero no es valido (no es un entero)")

Chi(input())
```
