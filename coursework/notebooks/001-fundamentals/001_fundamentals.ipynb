## Ejercicio 1

Con numpy, genera un array de 50 números aleatorios entre 0 y 1.

Calcula la media, la mediana y la desviación estándar de ese array.

Redondea los resultados a dos decimales e imprime cada valor.


```python
import numpy as np

random_generator = np.random.default_rng()
array = random_generator.random(50)
print("Array:\n", array.round(2), "\n")

print("Media:", f"{array.mean():.2f}")
print("Mediana:", f"{np.median(array):.2}")
print("Desviación estándar:", f"{array.std():.2f}")
```

## Ejercicio 2

Crea una lista con los números del 1 al 20.

Usando una comprensión de listas, genera otra lista con el cuadrado de cada número.

Filtra la lista para quedarte únicamente con los números cuadrados mayores a 100.


```python
nums = [i for i in range(1, 21)]
print("Lista del 1 al 20:\n", nums)

squares = [n**2 for n in nums]
print("Cuadrados de cada número:\n", squares)

squares_over_100 = [s for s in squares if s > 100]
print("Cuadrados mayores a 100:\n", squares_over_100)
```

## Ejercicio 4

Define una función que reciba un número entero positivo y devuelva:

"Par" si el número es par,

"Impar" si es impar,

Añade un mensaje de error si el número no es válido.


```python
def even_or_odd(number):
    if not isinstance(number, int) or number <= 0:
        return "Número no valido"
    return "Par" if number % 2 == 0 else "Impar"


print(even_or_odd(2))
```

## Ejercicio 5

Elabora un programa que sea capaz de calcular el IMC (Índice de masa corporal) de una persona. Los datos deberán ser ingresados por el usuario.

El IMC se calcula dividiendo los kilogramos de peso de una persona por el cuadrado de su estatura en metros (IMC = peso [kg]/ (estatura^2) [m2]).


```python
def imc(peso, estatura):
    return f"Tu IMC es {peso / estatura**2:.3f}"


peso = float(input("Introduce tu peso en kilogramos: "))
estatura = float(input("Introduce tu estatura en metros: "))

imc(peso, estatura)
```
